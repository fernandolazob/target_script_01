```python
query=f'''
    select *  from CRONOX.dbo.borrar_prestamo_dinner
    where fecha_envio='2026-05-27'
    and title='PPD PLUS'

'''
df_list = obtener_tabla_sql(spark,query,server_sa,user_sa,pwd_sa,db_sa)
df_list=df_list.join(df_venta,['vendor_lead_code'],'left')
df_list=df_list.withColumn('retiro',when(F.col('venta')==1,F.lit('venta')).otherwise(F.col('retiro'))).drop('venta')
df_list = df_list.withColumn(
    "tea_oferta",
    F.when(F.col("tea_ppd").isNull(), "no_aplica")
     .when(F.col("tea_formato") < F.col("tea_ppd"), "tea < tea_anterior")
     .when(F.col("tea_formato") > F.col("tea_ppd"), "tea > tea_anterior")
     .otherwise("tea = tea_ppd")
)

df_list = df_list.withColumn(
    "dif_tasa_formato",
    F.when(
        F.col("tea_ppd").isNull(),
        None
    ).otherwise(
        F.col("tea_ppd") - F.col("tea_formato")
    )
)

df_list = df_list.withColumn(
    "seg_comparativo_tasa_anterior",
    F.when(F.col("dif_tasa_formato").isNull(), "8. SIN DATO")
     .when(F.col("dif_tasa_formato") <= -0.10, "7. CAIDA MUY ALTA")
     .when(F.col("dif_tasa_formato") <= -0.05, "6. CAIDA ALTA")
     .when(F.col("dif_tasa_formato") < 0, "5. CAIDA BAJA")
     .when(F.col("dif_tasa_formato") == 0, "4. SE MANTIENE")
     .when(F.col("dif_tasa_formato") <= 0.02, "3. MEJORA BAJA")
     .when(F.col("dif_tasa_formato") <= 0.05, "2. MEJORA MEDIA")
     .otherwise("1. MEJORA ALTA")
)


df_list = df_list.withColumn(
    "dif_tasa",
    F.when(
        F.col("tasa_app").isNull(),
        None
    ).otherwise(
        F.col("tea_formato") - F.col("tasa_app")
    )
)

df_list = df_list.withColumn(
    "seg_comparativo_tasa_app",
    F.when(F.col("dif_tasa").isNull(), "8. SIN DATO")

     .when(F.col("dif_tasa") < -0.05, "7. APP CAIDA ALTA")
     .when(F.col("dif_tasa") < 0, "6. APP CAIDA BAJA")

     .when(F.col("dif_tasa") == 0, "5. SE MANTIENE")
     .when(F.col("dif_tasa") <= 0.015, "4. APP MEJORA BAJA")
     .when(F.col("dif_tasa") <= 0.030, "3. APP MEJORA MEDIA")

     .when(F.col("dif_tasa") <= 0.050, "2. APP MEJORA ALTA")
     .otherwise("1. APP MEJORA MUY ALTA")
)
df_list = df_list.withColumn(
    "dif_prestamo_actual",
    F.when(
        F.col("tasa_macro").isNull(),
        None
    ).otherwise(
        F.col("tea_formato") - F.col("tasa_macro")
    )
)

df_list = df_list.withColumn(
    "seg_comparativo_tasa_macro",
    F.when(F.col("dif_prestamo_actual").isNull(), "8. SIN DATO")
     .when(F.col("dif_prestamo_actual") < -0.10, "7. MACRO CAIDA MUY ALTA")
     .when(F.col("dif_prestamo_actual") < -0.05, "6. MACRO CAIDA ALTA")
     .when(F.col("dif_prestamo_actual") < 0, "5. MACRO CAIDA BAJA")
     .when(F.col("dif_prestamo_actual") == 0, "4. SE MANTIENE")
     .when(F.col("dif_prestamo_actual") <= 0.02, "3. MACRO MEJORA BAJA")
     .when(F.col("dif_prestamo_actual") <= 0.05, "2. MACRO MEJORA MEDIA")
     .otherwise("1. MACRO MEJORA ALTA")
)

print(df_list.columns)
print(df_list.count())
```

```python
from pyspark.sql import functions as F
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LogNorm
df_plot = (
    df_filtrado.groupBy("seg_edad", "prioridad_inicial")
    .count()
)

pdf = df_plot.toPandas()

pdf["prioridad_inicial"] = pdf["prioridad_inicial"].fillna("NULL").astype(str)
pdf["seg_edad"] = pdf["seg_edad"].fillna("NULL").astype(str)

tabla = pdf.pivot(index="seg_edad", columns="prioridad_inicial", values="count").fillna(0)
tabla = tabla.sort_index()
tabla = tabla.reindex(sorted(tabla.columns, key=lambda x: str(x)), axis=1)

fig, ax = plt.subplots(figsize=(10, 6))

# im = ax.imshow(tabla.values, aspect="auto")
im = ax.imshow(
    tabla.values,
    aspect="auto",
    cmap="tab20b",
    norm=LogNorm(vmin=1, vmax=tabla.values.max())
)

ax.set_xticks(np.arange(-.5, len(tabla.columns), 1), minor=True)
ax.set_yticks(np.arange(-.5, len(tabla.index), 1), minor=True)
ax.grid(which="minor", color="gray", linestyle='-', linewidth=0.3)
ax.tick_params(which="minor", bottom=False, left=False)

ax.set_title("Conteo de seg_edad vs prioridad_inicial")
ax.set_xlabel("prioridad_inicial")
ax.set_ylabel("seg_edad")

ax.set_xticks(np.arange(len(tabla.columns)))
ax.set_xticklabels(tabla.columns)
ax.set_yticks(np.arange(len(tabla.index)))
ax.set_yticklabels(tabla.index)

# mostrar etiquetas dentro de cada celda
for i in range(tabla.shape[0]):
    for j in range(tabla.shape[1]):
        ax.text(j, i, int(tabla.iloc[i, j]), ha="center", va="center")

plt.colorbar(im, ax=ax, label="Cantidad")
plt.tight_layout()
plt.show()
```
