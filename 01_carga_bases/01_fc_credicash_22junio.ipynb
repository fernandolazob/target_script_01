```python
from pyspark.sql import functions as F

cols_finales = [
    "CAMPANA",
    "FLG_PEARS",
    "FRESCURA",
    "OFERTA_CREDICASH",
    "OFERTA_ELECTRO",
    "PROPENSION_CREDICASH",
    "PROPENSION_ELECTRO",
    "Region",
    "Semaforo",
    "TASA_CREDICASH",
    "NUMERO_DOCUMENTO",
    "PRODUCTO_EXTERNO",
    "mes"
]

def limpiar_base(df):
    df = df.select(*cols_finales)

    cols_string = [
        "CAMPANA",
        "FLG_PEARS",
        "FRESCURA",
        "OFERTA_CREDICASH",
        "OFERTA_ELECTRO",
        "PROPENSION_CREDICASH",
        "PROPENSION_ELECTRO",
        "Region",
        "Semaforo",
        "TASA_CREDICASH",
        "NUMERO_DOCUMENTO",
        "PRODUCTO_EXTERNO"
    ]

    for c in cols_string:
        df = df.withColumn(
            c,
            F.when(
                F.col(c).isNull() |
                (F.trim(F.col(c).cast("string")) == "") |
                (F.lower(F.trim(F.col(c).cast("string"))) == "nan") |
                (F.lower(F.trim(F.col(c).cast("string"))) == "none") |
                (F.lower(F.trim(F.col(c).cast("string"))) == "null"),
                F.lit("no_aplica")
            ).otherwise(F.trim(F.col(c).cast("string")))
        )

    cols_quitar_decimal = [
        "FRESCURA",
        "OFERTA_CREDICASH",
        "OFERTA_ELECTRO",
        "PROPENSION_CREDICASH",
        "PROPENSION_ELECTRO",
        "TASA_CREDICASH"
    ]

    for c in cols_quitar_decimal:
        df = df.withColumn(
            c,
            F.when(F.col(c) == "no_aplica", F.col(c))
             .otherwise(F.regexp_replace(F.col(c), r"\.0$", ""))
        )

    df = df.withColumn("mes", F.col("mes").cast("int"))

    return df
```

```python
def analizar_dimension(df, columna):
    return (
        df
        .groupBy(columna)
        .agg(
            F.countDistinct(F.when(F.col("mes") == 5, F.col("NUMERO_DOCUMENTO"))).alias("Q_Mayo"),
            F.countDistinct(F.when(F.col("mes") == 6, F.col("NUMERO_DOCUMENTO"))).alias("Q_Junio")
        )
        .withColumn("Dif_Q", F.col("Q_Junio") - F.col("Q_Mayo"))
        .withColumn(
            "VarPct_Q",
            F.when(
                F.col("Q_Mayo") != 0,
                F.round((F.col("Dif_Q") / F.col("Q_Mayo")) * 100, 2)
            ).otherwise(None)
        )
        .orderBy(F.col("Dif_Q").asc())
    )
```

```python
from pyspark.sql import functions as F

def comparar_por_columna(df, columna):
    df_tmp = (
        df.groupBy(columna, "mes")
        .agg(F.countDistinct("NUMERO_DOCUMENTO").alias("registros"))
        .groupBy(columna)
        .pivot("mes", [5, 6])
        .agg(F.first("registros"))
        .fillna(0)
    )

    return (
        df_tmp
        .withColumnRenamed("5", "mayo")
        .withColumnRenamed("6", "junio")
        .withColumn("diferencia", F.col("junio") - F.col("mayo"))
        .withColumn(
            "variacion_pct",
            F.when(F.col("mayo") != 0, F.round((F.col("diferencia") / F.col("mayo")) * 100, 2))
             .otherwise(None)
        )
        .orderBy(F.col("diferencia").asc())
    )
```

```python
from pyspark.sql import functions as F

def comparar_por_columna(df, columna):
    return (
        df.groupBy(columna, "mes")
        .agg(F.countDistinct("NUMERO_DOCUMENTO").alias("registros"))
        .groupBy(columna)
        .agg(
            F.sum(F.when(F.col("mes") == 5, F.col("registros")).otherwise(0)).alias("mayo"),
            F.sum(F.when(F.col("mes") == 6, F.col("registros")).otherwise(0)).alias("junio")
        )
        .withColumn("diferencia", F.col("junio") - F.col("mayo"))
        .withColumn(
            "variacion_pct",
            F.when(F.col("mayo") != 0, F.round((F.col("diferencia") / F.col("mayo")) * 100, 2))
             .otherwise(None)
        )
        .orderBy(F.col("diferencia").asc())
    )

df_base = df_base.withColumn(
    "SEGMENTO_PRODUCTO",
    F.when(
        (F.col("OFERTA_CREDICASH").isNotNull()) & (F.col("OFERTA_ELECTRO").isNotNull()),
        "CREDICASH / ELECTRO"
    )
    .when(
        (F.col("OFERTA_CREDICASH").isNotNull()) & (F.col("OFERTA_ELECTRO").isNull()),
        "CREDICASH"
    )
    .when(
        (F.col("OFERTA_CREDICASH").isNull()) & (F.col("OFERTA_ELECTRO").isNotNull()),
        "ELECTRO"
    )
    .otherwise("SIN OFERTA")
)

df_general = (
    df_base.groupBy("mes")
    .agg(F.countDistinct("NUMERO_DOCUMENTO").alias("registros"))
)

df_comp_general = (
    df_general.groupBy()
    .agg(
        F.sum(F.when(F.col("mes") == 5, F.col("registros")).otherwise(0)).alias("mayo"),
        F.sum(F.when(F.col("mes") == 6, F.col("registros")).otherwise(0)).alias("junio")
    )
    .withColumn("diferencia", F.col("junio") - F.col("mayo"))
    .withColumn("variacion_pct", F.round((F.col("diferencia") / F.col("mayo")) * 100, 2))
)

df_segmento = comparar_por_columna(df_base, "SEGMENTO_PRODUCTO")
df_propension = comparar_por_columna(df_base, "PROPENSION_CREDICASH")
df_oferta = comparar_por_columna(df_base, "OFERTA_CREDICASH")
df_tasa = comparar_por_columna(df_base, "TASA_CREDICASH")
df_region = comparar_por_columna(df_base, "Region")
df_semaforo = comparar_por_columna(df_base, "Semaforo")
df_frescura = comparar_por_columna(df_base, "FRESCURA")
```
